```python
# Install package (run only once)
!pip install statsmodels

# Import Libraries
import pandas as pd
import numpy as np
from statsmodels.stats.weightstats import ztest

# Load Dataset
uci_diabetes = pd.read_csv('/content/uci_diabetes.csv')

# Display first few rows
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

# Perform Z-Test
z_stat, p_value = ztest(uci_diabetes["Glucose"], value=100)

# Display Results
print("\nZ-Test Results")
print("----------------------")
print(f"Z-Statistic : {z_stat:.4f}")
print(f"P-Value     : {p_value:.4f}")

# Interpretation
alpha = 0.05

if p_value < alpha:
    print("\nReject the Null Hypothesis")
    print("The mean Glucose level is significantly different from 100.")
else:
    print("\nFail to Reject the Null Hypothesis")
    print("No significant difference in the mean Glucose level.")
```
