```python
import pandas as pd
from google.colab import files

# Read CSV File
text_df = pd.read_csv('/content/Google_data.csv')

# Create Excel DataFrame
excel_df = pd.DataFrame({
    "Product": ["Laptop", "Smartphone", "Tablet", "Headphones"],
    "Price": [1000, 800, 500, 100],
    "Quantity": [5, 8, 10, 15]
})

# Read Web Data
web_df = pd.read_csv("https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv")

# Display Data
print("TEXT FILE")
print(text_df.head())

print("\nEXCEL FILE")
print(excel_df.head())

print("\nWEB DATA")
print(web_df.head())

# Handle Missing Values
text_df = text_df.ffill()
excel_df = excel_df.bfill()
web_df = web_df.dropna()

# Save Files
text_df.to_csv('/content/processed_text.csv', index=False)
excel_df.to_excel('/content/processed_excel.xlsx', index=False)

print("\nFiles saved successfully!")

# Download Files
files.download('/content/processed_text.csv')
files.download('/content/processed_excel.xlsx')
```
