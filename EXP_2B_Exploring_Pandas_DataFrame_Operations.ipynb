```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'existing_column': [20, 55, 70, 90, 45],
    'another_column': [120, 80, 60, 95, 110],
    'category_column': ['A', 'B', 'A', 'B', 'A'],
    'numeric_column': [100, 200, 150, 250, 180],
    'column1': ['X', 'Y', 'Z', 'P', 'Q'],
    'column2': [10, 20, 30, 40, 50]
})

print("First 5 rows:\n", df.head())
print("Last 5 rows:\n", df.tail())

df.info()

print("\nSummary Statistics:\n", df.describe())

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

# Create a new column
df['new_column'] = df['existing_column'] * 2

# Series operations
series = df['existing_column']
print("\nSeries addition:\n", series + 10)

# Filter rows
filtered_df = df[(df['existing_column'] > 50) &
                 (df['another_column'] < 100)]
print("\nFiltered DataFrame:\n", filtered_df)

# Grouping
grouped = df.groupby('category_column')['numeric_column'].mean()
print("\nGrouped Mean:\n", grouped)

# Sorting
df_sorted = df.sort_values(by='numeric_column', ascending=False)
print("\nSorted DataFrame:\n", df_sorted)

# Boolean masking
masked_df = df[df['numeric_column'] > df['numeric_column'].median()]
print("\nMasked DataFrame:\n", masked_df)

# Remove duplicates and missing values
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# Subset
subset_df = df[['column1', 'column2']]

# Save CSV
subset_df.to_csv('filtered_data.csv', index=False)

# Summary
print("\nTotal Sum:", df['numeric_column'].sum())
print("Mean:", df['numeric_column'].mean())
print("Standard Deviation:", df['numeric_column'].std())
```
