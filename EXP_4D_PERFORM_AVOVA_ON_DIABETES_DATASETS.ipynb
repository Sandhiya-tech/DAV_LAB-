```python
# Import Libraries
import pandas as pd
import numpy as np
from scipy.stats import f_oneway

# Load the Datasets
uci_diabetes = pd.read_csv('/content/uci_diabetes.csv')
pima_diabetes = pd.read_csv('/content/pima_diabetes.csv')

# Display first few rows
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())

print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())

# Select Relevant Numerical Columns
numerical_columns = ["Glucose", "BloodPressure", "BMI"]

# Perform One-Way ANOVA
anova_results = {}

for col in numerical_columns:
    f_stat, p_value = f_oneway(
        uci_diabetes[col],
        pima_diabetes[col]
    )

    anova_results[col] = {
        "F-statistic": round(f_stat, 6),
        "P-value": round(p_value, 6)
    }

# Convert Results to DataFrame
anova_df = pd.DataFrame(anova_results).T

# Display Results
print("\nANOVA Results:")
print(anova_df)

# Interpretation
print("\nInterpretation:")
for col in numerical_columns:
    if anova_df.loc[col, "P-value"] < 0.05:
        print(f"{col}: Significant difference exists (Reject H0)")
    else:
        print(f"{col}: No significant difference (Fail to Reject H0)")
```
